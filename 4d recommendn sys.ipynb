```python
import pandas as pd
import numpy as np
import re
from gensim.models import Word2Vec
from sklearn.metrics.pairwise import cosine_similarity
```

```python
data = pd.read_csv("twitter_dataset.csv")

def clean_text(text):
    text = str(text).lower()
    text = re.sub(r'http\S+|www\S+', '', text)
    text = re.sub(r'[^a-z\s]', '', text)
    return text.split()

data['tokens'] = data['Text'].apply(clean_text)

print(data['tokens'].head())
```

```python
sentences = data['tokens'].tolist()

model = Word2Vec(
    sentences=sentences,
    vector_size=100,
    window=5,
    min_count=2,
    sg=0,
    epochs=10
)

print("Vocabulary size:", len(model.wv))
```

```python
def get_vector(words):
    vectors = []

    for word in words:
        if word in model.wv:
            vectors.append(model.wv[word])

    if len(vectors) == 0:
        return np.zeros(100)

    return np.mean(vectors, axis=0)


document_vectors = np.array([
    get_vector(words) for words in data['tokens']
])

print(document_vectors.shape)
```

```python
user_input = input("Enter your search query: ")

query_words = clean_text(user_input)

query_vector = get_vector(query_words)

print("Query vector shape:", query_vector.shape)
```

```python
similarities = cosine_similarity(
    [query_vector],
    document_vectors
)[0]

print("Similarity calculated successfully")
```

```python
top_indices = np.argsort(similarities)[::-1][:5]

print("Recommended Tweets:\n")

for i in top_indices:
    print(data['Text'].iloc[i])
    print("Similarity:", similarities[i])
    print()
```
