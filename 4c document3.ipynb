```python
import re
import numpy as np
from gensim.models import Word2Vec
```

```python
with open("document3.txt", "r", encoding="utf-8") as file:
    text = file.read()

print(text[:500])
```

```python
text = text.lower()

text = re.sub(r'[^a-z\s]', '', text)

words = text.split()

print(words[:30])
```

```python
sentences = [words]

print("Total words:", len(words))
```

```python
model = Word2Vec(
    sentences=sentences,
    vector_size=100,
    window=5,
    min_count=1,
    sg=0,
    epochs=10
)

print("Vocabulary Size:", len(model.wv))
```

```python
word = words[0]

print("Word:", word)
print("Embedding:", model.wv[word])
```

```python
word = words[0]

print("Similar words to:", word)

print(model.wv.most_similar(word, topn=5))
```

```python
word1 = words[0]
word2 = words[1]

similarity = model.wv.similarity(word1, word2)

print("Similarity:", similarity)
```
